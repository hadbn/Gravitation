```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
fig, ax = plt.subplots()


ax.axis('equal')
ax.set(xlim=[-1, 30], ylim=[-1,30])
G=5
T=0.005
n=3
P=np.random.uniform(5,25,(n,2))      #positions initiales
V=np.zeros((n,2))                   #vitesses initiales nulles
M=np.random.uniform(10,50,n)        #masses des éléments
print("n =", n)
print("Nombre de positions :", len(P))
print("Nombre de vitesses :", len(V))
print("Nombre de masses :", len(M))
print("Positions initiales :\n", P)

def dist(L1,L2):              #calcul de la distance entre deux éléments
    return(np.sum((L1-L2)**2))

#def force(i,j,P,M): #force exercée par j sur i
#   return(G*(P[j]-P[i])*M[j]/(dist(P[i],P[j])**(3/2)))

#def nouvelle_vitesse(P,V,M,n):        #calcul d'un vecteur V qui contient les vitesses de chaque élément
#    for i in range(len(V)):
#      V[i]=V[i]+T*sum(force(i,j,P,M) for j in range(n) if j!=i )
#   return(V)

#Pour avoir V directement avec numpy
def nvelle_vitesse(P,V,M,n):
    V = V + T * G * np.array([sum(M[j] * (P[j] - P[i]) / (dist(P[i], P[j])**(3/2)) for j in range(n) if j != i) for i in range(n)])
    return V

def nouvelle_position(P,V,M,n):   #calcul de la nouvelle position à l'aide de la vitesse par méthode d'Euler
    V=nvelle_vitesse(P,V,M,n)
    return(P+V*T,V)      

    
  
    
scat = ax.scatter(P[:,0], P[:,1])  # on récupère les positions initiales
print("Nombre de points affichés :", len(scat.get_offsets()))

print(n, P.shape, V.shape, M.shape)
print(scat.get_offsets().shape)

def animate(t):
    # une variable globale est une variable utilisée dans une fonction mais dont la modification de la valeur a une portée globale (donc extérieure à la fonction)

    global P
    global V
    P,V = nouvelle_position(P,V,M,n)

    # update the scatter plot:
    # le np.stack sert ici à mettre les positions dans la bonne shape
    data = np.stack(P).T
    scat.set_offsets(data)
    return scat,

ani = animation.FuncAnimation(fig,animate,frames=300,interval=50,blit=True,cache_frame_data=False)  #la ligne fournie en exemple ne donnait pas d'animation sur jupyter lab, on a donc demandé à une ia de générer cette ligne de code pour avoir l'animation
plt.close(fig)                                          #même chose
from IPython.display import HTML, display               #même chose
display(HTML(ani.to_jshtml()))                          #même chose
```
